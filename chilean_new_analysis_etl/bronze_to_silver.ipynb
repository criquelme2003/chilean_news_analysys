```python
# %load_ext autoreload
# %autoreload 2
from environment import SILVER_TABLES,BRONZE_TABLES
```

```python
spark.sql("CREATE SCHEMA IF NOT EXISTS chilean_news.checkpoints")
spark.sql("CREATE SCHEMA IF NOT EXISTS chilean_news.silver_tables")
spark.sql("CREATE VOLUME IF NOT EXISTS chilean_news.checkpoints.silver")

```

```python
import pyspark.sql.functions as F
```

### Stream for raw tables 


```python

rn_bronze = BRONZE_TABLES["raw_news"]
rn_silver = SILVER_TABLES["raw_news"]


rstr = spark.read\
    .table(rn_bronze)\
    .where((F.col("medio").isNotNull()) & (F.trim(F.col("medio")) != ""))\
    .where((F.col("id").isNotNull()) & (F.trim(F.col("id")) != ""))\
    .where((F.col("title").isNotNull()) & (F.trim(F.col("title")) != ""))\
    .select("id","author","medio","title","link")

rstr.write.mode("overwrite")\
    .option("overwriteSchema","true")\
    .partitionBy("medio")\
    .saveAsTable(rn_silver)


br = spark.read.table(rn_bronze)
sr = spark.read.table(rn_silver)


print(f"Bronze table: {br.count()} rows")
print(f"Silver table: {sr.count()} rows")
```

```python

rn_bronze = BRONZE_TABLES["news_diario_financiero"]
rn_silver = SILVER_TABLES["news_diario_financiero"]

rstr = spark.read\
    .table(rn_bronze)\
    .where((F.col("link").isNotNull()) & (F.trim(F.col("link")) != ""))\
    .where((F.col("texto_completo").isNotNull()) & (F.trim(F.col("texto_completo")) != ""))\
    .where((F.col("titulo").isNotNull()) & (F.trim(F.col("titulo")) != ""))\
    .select("autor","titulo","texto_completo","link",F.try_to_date("fecha_publicacion").alias("fecha_publicacion"))\
    .where(F.col("fecha_publicacion").isNotNull())

rstr.write\
    .mode("overwrite")\
    .partitionBy("autor")\
    .option("checkpointLocation","/Volumes/chilean_news/checkpoints/silver/diario_financiero")\
    .saveAsTable(rn_silver)


br = spark.read.table(rn_bronze)
sr = spark.read.table(rn_silver)

print(f"Bronze table: {br.count()} rows")
print(f"Silver table: {sr.count()} rows")
```
